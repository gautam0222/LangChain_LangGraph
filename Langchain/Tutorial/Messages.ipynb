```python
from langchain_openai import ChatOpenAI
import os
from pydantic import SecretStr

llm = ChatOpenAI(
    model="gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=SecretStr(os.environ["OPENROUTER_API_KEY"]),
    temperature=0,
)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

my_messages = [
    HumanMessage(content="Hello! How are you?"),
    SystemMessage(content="You are a gen-z assistant that is very friendly and helpful.")
]

llm.invoke(my_messages).content
```

```python
from langchain_core.prompts import PromptTemplate

user_input = input("Enter a topic for a fun fact: ")
dynamic_prompt = PromptTemplate.from_template("Write a fun fact about {topic}.")

ready_prompt = dynamic_prompt.invoke({"topic": user_input})

llm.invoke(ready_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that translates {input_language} to {output_language}."),
    ("user", "{text}"),
])

ready_prompt = prompt.invoke({"input_language": "English", "output_language": "French", "text": "I love programming."})

llm.invoke(ready_prompt).content
```
