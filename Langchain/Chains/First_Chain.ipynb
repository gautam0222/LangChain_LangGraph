```python
from langchain_openai import ChatOpenAI
import os
from pydantic import SecretStr

llm = ChatOpenAI(
    model="gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=SecretStr(os.environ["OPENROUTER_API_KEY"])
)
```

## First Chain

```python
#Task 1
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "{text}"),
])
```

```python
#Task 2
llm_openai = ChatOpenAI(
    model="gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=SecretStr(os.environ["OPENROUTER_API_KEY"]),
    temperature=0.7
)
```

```python
#Task 3
from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()
```

## Manual Invocation

```python
template = prompt_template.invoke({"text": "Write a fun fact about programming."})

result = llm.invoke(template)

final_result = result.content

print(final_result)
```

## Chain Invocation

```python
chain = prompt_template | llm_openai | str_parser

chain.invoke({"text": "Write a fun fact about programming."})
```

## Runnable Sequence

```python
from langchain_core.runnables import RunnableSequence

chain_sequence = RunnableSequence(prompt_template, llm_openai, str_parser)

chain_sequence.invoke({"text": "Write a fun fact about programming."})
```
