```python
from langchain_openai import ChatOpenAI
import os
from pydantic import SecretStr

llm = ChatOpenAI(
    model="gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=SecretStr(os.environ["OPENROUTER_API_KEY"]),
    temperature=0,
)   
llm
```

```python
from pydantic import BaseModel, Field
from typing import List, TypedDict, Literal

class llm_schema(BaseModel):
    joke: str = Field(..., description="The joke generated by the LLM.")
    funny_flag: Literal["Funny", "Not Funny"] = Field(..., description="Indicates whether the joke is funny or not.")
    feedback: str = Field(..., description="Feedback on the joke.")

llm_with_schema=llm.with_structured_output(llm_schema)

llm_with_schema.invoke("Tell me a joke about programming.")
```

```python
class graph_schema(TypedDict):

    topic: str
    joke: str
    funny_flag: str
    feedback: str
    max_iterations: int
```

```python
def generate_node(state: graph_schema) -> graph_schema:

    topic=state["topic"]

    if state["feedback"]:
        response=llm_with_schema.invoke(f"Tell me a joke about {topic}. The previous joke was: {state['joke']}. The feedback was: {state['feedback']}. Please provide a new joke.")
    else:
        response=llm_with_schema.invoke(f"Tell me a joke about {topic}.")

    state["joke"]=response.joke
    return state
```

```python
from langchain_core.prompts import ChatPromptTemplate

def evaluate_node(state: graph_schema) -> graph_schema:

    joke=state["joke"]
    iteration=state["max_iterations"]

    prompt=ChatPromptTemplate.from_messages([
        ("system","You are a joke evaluator. You will be given a joke and you will evaluate it based on how funny it is. You will provide feedback on the joke and indicate whether it is funny or not."),
        ("user",f"Here is the joke: {joke}. Please evaluate the joke and provide feedback on how funny it is. Indicate whether the joke is funny or not by responding with either 'Funny' or 'Not Funny'.")
    ])

    chain= prompt | llm_with_schema
    response=chain.invoke({})

    state["funny_flag"]=response.funny_flag
    state["feedback"]=response.feedback
    state["max_iterations"]=iteration+1

    return state
```

```python
def check_iterations(state: graph_schema) -> str:

    iteration=state["max_iterations"]

    if iteration>=3 or state["funny_flag"]=="Funny":
        return "end"
    else:
        return "generate_node"
```

```python
from langgraph.graph import StateGraph, START, END

graph=StateGraph(graph_schema)
graph.add_node("generate_node", generate_node)
graph.add_node("evaluate_node", evaluate_node)

graph.add_edge(START, "generate_node")
graph.add_edge("generate_node", "evaluate_node")
graph.add_conditional_edges("evaluate_node", check_iterations, {
    "generate_node": "generate_node",
    "end": END
})

evaluator_graph=graph.compile()
```

```python
from IPython.display import display

display(evaluator_graph)
```

```python
for chunk in evaluator_graph.stream(
    {
        "topic": "programming",
        "joke": "",
        "funny_flag": "",
        "feedback": "",
        "max_iterations": 0
    }
):
    print(chunk)
```
